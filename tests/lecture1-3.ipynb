## einops2

```python
x = torch.ones(3, 4)  # seq1 hidden
y = torch.ones(4, 3)  # hidden seq2
# Old way
z = x @ y   # seq1 seq2
# New (einops) way
z = einsum(x, y, "seq1 hidden1, hidden1 seq2 -> seq1 seq2")
```
